## ⚡ Step 3: Run Isolated Code Execution via CLI & Python SDK

```python
# Run code sandbox via CLI
!thinkdome run "import sys; print('Hello from ThinkDome CLI Sandbox!'); print('Python version:', sys.version)"
```

```python
import asyncio
from thinkdome.sandbox.core.service import SandboxExecutionService, ExecuteRequest

async def run_sandbox_demo():
    service = SandboxExecutionService()
    request = ExecuteRequest(
        code="""
import math
result = [math.sqrt(x) for x in range(1, 6)]
print(f"Computed inside ThinkDome Sandbox: {result}")
""",
        language="python",
        security_profile="HIGH_SECURITY"
    )
    response = await service.execute(request)
    print("--- Sandbox Execution Results ---")
    print("Exit Code:", response.exit_code)
    print("Stdout:", response.stdout)

await run_sandbox_demo()
```

## 🌐 Step 4: Host Ephemeral Web Preview (`host_html`) with Custom 24h Expiry

```python
import json
from thinkdome.sandbox.tools.execution_tools import HostHtmlTool

async def test_web_preview():
    tool = HostHtmlTool()
    html_content = """<!DOCTYPE html>
<html>
<head>
  <title>ThinkDome Colab Preview</title>
  <style>
    body { font-family: sans-serif; background: #ffffff; color: #0f172a; padding: 40px; text-align: center; }
    .card { border: 1px solid #e2e8f0; border-radius: 12px; padding: 30px; box-shadow: 0 10px 25px rgba(0,0,0,0.05); max-width: 500px; margin: 0 auto; }
    h1 { color: #0284c7; }
  </style>
</head>
<body>
  <div class="card">
    <h1>ThinkDome Ephemeral Gateway</h1>
    <p>Hosted from Google Colab with 24-hour custom TTL.</p>
    <p><b>TTL Expiry:</b> 86400 Seconds</p>
  </div>
</body>
</html>"""
    
    result = await tool.execute({
        "html": html_content,
        "ttl_sec": 86400, # 24 Hours Custom Expiry
        "filename": "index.html",
        "site_name": "colab_demo"
    })
    print(json.dumps(json.loads(result), indent=2))

await test_web_preview()
```

## 🚀 Step 5: Launch ThinkDome API Server & Web Console in Colab

```python
# Start ThinkDome API Server in background thread and serve UI inside Colab
import threading, time
import uvicorn
from google.colab import output

def start_thinkdome():
    uvicorn.run("thinkdome.api.server:create_app", factory=True, host="127.0.0.1", port=8000)

# Launch server background thread
server_thread = threading.Thread(target=start_thinkdome, daemon=True)
server_thread.start()
time.sleep(3)

print("✅ ThinkDome API Server running on port 8000!")
print("🔗 Click below to open the Web Console window:")
output.serve_kernel_port_as_window(8000)
```
